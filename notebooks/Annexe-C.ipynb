# Annexe C — code Sage des figures 3D

Code SageMath d'origine des figures 3D de l'annexe C (cylindres et surfaces quadriques), remplacées dans le manuel par des figures Asymptote.
Chaque cellule est autonome : on peut la copier dans [SageMathCell](https://sagecell.sagemath.org) pour obtenir une figure qu'on peut faire pivoter.

Les numéros de figures sont ceux de la version compilée au moment de l'extraction (octobre 2026).


## Section C.1 — Cylindres


### Figure C.1.3

Le cylindre parabolique $z=x^2$.


```sage
var('x,y')
def macouleur(x,y):
    return float(x^2/4)
S = plot3d(x^2, (x,-2,2), (y,-2,2), color=(macouleur, colormaps.Spectral), mesh=True, opacity=0.85, frame=False)
S += arrow3d((0,0,0), (2.3,0,0), color="black")
S += text3d("x", (2.6,0,0), color="black", fontsize=24)
S += arrow3d((0,0,0), (0,2.3,0), color="black")
S += text3d("y", (0,2.6,0), color="black", fontsize=24)
S += arrow3d((0,0,0), (0,0,4.5), color="black")
S += text3d("z", (0,0,4.8), color="black", fontsize=24)
S
```

### Figure C.1.4

Un cylindre dont la base est une cardioïde.


```sage
var('t,z')
r(t) = 1 + cos(t)
def macouleur(t,z):
    return float((z+1)/2)
S = parametric_plot3d([r(t)*cos(t), r(t)*sin(t), z], (t,0,2*pi), (z,-1,1), color=(macouleur, colormaps.Spectral), mesh=True, opacity=0.85)
S += arrow3d((0,0,0), (2.3,0,0), color="black")
S += text3d("x", (2.6,0,0), color="black", fontsize=24)
S += arrow3d((0,0,0), (0,1.8,0), color="black")
S += text3d("y", (0,2.1,0), color="black", fontsize=24)
S += arrow3d((0,0,0), (0,0,1.5), color="black")
S += text3d("z", (0,0,1.8), color="black", fontsize=24, frame = False)
S
```

## Section C.2 — Surfaces quadriques


### Figure C.2.3

Un ellipsoïde.


```sage
var('u,v,t')
a,b,c = 1, 2.5, 1.5
def macouleur(u,v):
    z = c*cos(v)
    return float(0.5 + 0.25*(z/c))
S = parametric_plot3d([a*cos(u)*sin(v), b*sin(u)*sin(v), c*cos(v)], (u,0,2*pi), (v,0,pi), color=(macouleur, colormaps.YlOrBr), mesh=False, opacity=1, frame=False)
S += parametric_plot3d([a*cos(t), b*sin(t), 0], (t,0,2*pi), color="firebrick", thickness=1.5)
S += parametric_plot3d([a*cos(t), 0, c*sin(t)], (t,0,2*pi), color="seagreen", thickness=1.5)
S += parametric_plot3d([0, b*cos(t), c*sin(t)], (t,0,2*pi), color="steelblue", thickness=1.5)
s = sqrt(3)/2
S += parametric_plot3d([a*s*cos(t), b*s*sin(t), c/2], (t,0,2*pi), color="firebrick", thickness=1.5)
S += parametric_plot3d([a*s*cos(t), b/2, c*s*sin(t)], (t,0,2*pi), color="seagreen", thickness=1.5)
S += parametric_plot3d([a/2, b*s*cos(t), c*s*sin(t)], (t,0,2*pi), color="steelblue", thickness=1.5)
S += arrow3d((0,0,0), (a+0.5,0,0), color="black")
S += text3d("x", (a+0.8,0,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,b+0.5,0), color="black")
S += text3d("y", (0,b+0.8,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,0,c+0.5), color="black")
S += text3d("z", (0,0,c+0.8), color="black", fontsize=26, frame = False)
S
```

### Figure C.2.4

Un hyperboloïde à une nappe.


```sage
var('u,v,t')
a,b,c = 1, 2.5, 1.5
vmax = 1.1
def macouleur(u,v):
    z = c*sinh(v)
    zmax = c*sinh(vmax)
    return float(0.5 + 0.25*(z/zmax))
S = parametric_plot3d([a*cosh(v)*cos(u), b*cosh(v)*sin(u), c*sinh(v)], (u,0,2*pi), (v,-vmax,vmax), color=(macouleur, colormaps.YlOrBr), mesh=False, opacity=1, frame=False)
S += parametric_plot3d([a*cos(t), b*sin(t), 0], (t,0,2*pi), color="firebrick", thickness=1.5)
v2 = vmax/2
S += parametric_plot3d([a*cosh(v2)*cos(t), b*cosh(v2)*sin(t), c*sinh(v2)], (t,0,2*pi), color="firebrick", thickness=1.5)
S += parametric_plot3d([a*cosh(t), 0, c*sinh(t)], (t,-vmax,vmax), color="seagreen", thickness=1.5)
S += parametric_plot3d([-a*cosh(t), 0, c*sinh(t)], (t,-vmax,vmax), color="seagreen", thickness=1.5)
S += parametric_plot3d([0, b*cosh(t), c*sinh(t)], (t,-vmax,vmax), color="steelblue", thickness=1.5)
S += parametric_plot3d([0, -b*cosh(t), c*sinh(t)], (t,-vmax,vmax), color="steelblue", thickness=1.5)
xmax, ymax, zmax = a*cosh(vmax), b*cosh(vmax), c*sinh(vmax)
S += arrow3d((0,0,0), (xmax+0.5,0,0), color="black")
S += text3d("x", (xmax+0.8,0,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,ymax+0.5,0), color="black")
S += text3d("y", (0,ymax+0.8,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,0,zmax+0.5), color="black")
S += text3d("z", (0,0,zmax+0.8), color="black", fontsize=26, frame = False)
S
```

### Figure C.2.5

Un hyperboloïde à deux nappes.


```sage
var('u,v,t')
a,b,c = 1.5, 2.5, 1
vmax = 1.1
vcap = 1.2*vmax  # la nappe est tronquée à v = vcap (un peu au-delà de vmax)
def macouleur(u,v):
    return float(0.4 + 0.1*(v/vmax))
# la nappe du haut (z positif), puis celle du bas (z négatif)
S = parametric_plot3d([a*sinh(v)*cos(u), b*sinh(v)*sin(u), c*cosh(v)], (u,0,2*pi), (v,0,vcap), color=(macouleur, colormaps.YlOrBr), mesh=False, opacity=0.85, frame=False)
S += parametric_plot3d([a*sinh(v)*cos(u), b*sinh(v)*sin(u), -c*cosh(v)], (u,0,2*pi), (v,0,vcap), color=(macouleur, colormaps.YlOrBr), mesh=False, opacity=0.85, frame=False)
# trace horizontale (z = cste) au sommet de chaque nappe : une ellipse
S += parametric_plot3d([a*sinh(vmax)*cos(t), b*sinh(vmax)*sin(t), c*cosh(vmax)], (t,0,2*pi), color="firebrick", thickness=1.5)
S += parametric_plot3d([a*sinh(vmax)*cos(t), b*sinh(vmax)*sin(t), -c*cosh(vmax)], (t,0,2*pi), color="firebrick", thickness=1.5)
# trace verticale dans le plan y=0 (donc y = cste = 0) : une hyperbole, sur les deux nappes
S += parametric_plot3d([a*sinh(t), 0, c*cosh(t)], (t,-vcap,vcap), color="seagreen", thickness=1.5)
S += parametric_plot3d([a*sinh(t), 0, -c*cosh(t)], (t,-vcap,vcap), color="seagreen", thickness=1.5)
# trace verticale dans le plan x=0 (donc x = cste = 0) : une hyperbole, sur les deux nappes
S += parametric_plot3d([0, b*sinh(t), c*cosh(t)], (t,-vcap,vcap), color="steelblue", thickness=1.5)
S += parametric_plot3d([0, b*sinh(t), -c*cosh(t)], (t,-vcap,vcap), color="steelblue", thickness=1.5)
# on répète les deux traces verticales précédentes, mais décalées à x = x0 et
# y = y0 strictement positifs (plutôt que par l'origine) ; tx, ty sont calculés
# pour que ces courbes s'arrêtent exactement au bord de la nappe tronquée
# (v = vcap), sans le dépasser
x0, y0 = a/2, b/2
ty = b*sqrt(sinh(vcap)^2 - (x0/a)^2)
tx = a*sqrt(sinh(vcap)^2 - (y0/b)^2)
S += parametric_plot3d([x0, t, c*sqrt(1+(x0/a)^2+(t/b)^2)], (t,-ty,ty), color="steelblue", thickness=1.5)
S += parametric_plot3d([x0, t, -c*sqrt(1+(x0/a)^2+(t/b)^2)], (t,-ty,ty), color="steelblue", thickness=1.5)
S += parametric_plot3d([t, y0, c*sqrt(1+(y0/b)^2+(t/a)^2)], (t,-tx,tx), color="seagreen", thickness=1.5)
S += parametric_plot3d([t, y0, -c*sqrt(1+(y0/b)^2+(t/a)^2)], (t,-tx,tx), color="seagreen", thickness=1.5)
xmax, ymax, zmax = a*sinh(vmax), b*sinh(vmax), c*cosh(vmax)
S += arrow3d((0,0,0), (xmax+0.5,0,0), color="black")
S += text3d("x", (xmax+0.8,0,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,ymax+0.5,0), color="black")
S += text3d("y", (0,ymax+0.8,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,0,1.2*zmax+0.5), color="black")
S += text3d("z", (0,0,1.2*zmax+0.8), color="black", fontsize=26, frame = False)
S
```

### Figure C.2.6

Un cône elliptique.


```sage
var('u,v,t')
a,b,c = 1, 2.5, 1.5
vmax = 1.3
def macouleur(u,v):
    z = c*v
    zmax = c*vmax
    return float(0.5 + 0.25*(z/zmax))
# le double cône, en entier (v négatif donne la nappe du bas)
S = parametric_plot3d([a*v*cos(u), b*v*sin(u), c*v], (u,0,2*pi), (v,-vmax,vmax), color=(macouleur, colormaps.YlOrBr), mesh=False, opacity=1, frame=False)
# traces horizontales (z = cste) : des ellipses, une au-dessus, une en-dessous du sommet
S += parametric_plot3d([a*vmax*cos(t), b*vmax*sin(t), c*vmax], (t,0,2*pi), color="firebrick", thickness=1.5)
v0 = vmax/2
S += parametric_plot3d([a*v0*cos(t), b*v0*sin(t), c*v0], (t,0,2*pi), color="firebrick", thickness=1.5)
S += parametric_plot3d([-a*v0*cos(t), -b*v0*sin(t), -c*v0], (t,0,2*pi), color="firebrick", thickness=1.5)
# génératrices (droites passant par le sommet) dans les plans y=0 et x=0
S += parametric_plot3d([a*t, 0, c*t], (t,-vmax,vmax), color="seagreen", thickness=1.5)
S += parametric_plot3d([0, b*t, c*t], (t,-vmax,vmax), color="steelblue", thickness=1.5)
# traces verticales décalées à x = x0 et y = y0 strictement positifs : ce ne sont
# plus des génératrices (elles ne passent pas par le sommet), mais des hyperboles.
# x0, y0 sont les coordonnées d'un même point p0 = (x0,y0,c*v0) du cône, à la
# hauteur v0 de l'ellipse ci-dessus : les trois traces (l'ellipse et les deux
# hyperboles) passent donc toutes par p0, comme demandé.
u0 = pi/4
x0 = a*v0*cos(u0)
y0 = b*v0*sin(u0)
ty = b*sqrt(vmax^2 - (x0/a)^2)
tx = a*sqrt(vmax^2 - (y0/b)^2)
S += parametric_plot3d([x0, t, c*sqrt((x0/a)^2+(t/b)^2)], (t,-ty,ty), color="steelblue", thickness=1.5)
S += parametric_plot3d([x0, t, -c*sqrt((x0/a)^2+(t/b)^2)], (t,-ty,ty), color="steelblue", thickness=1.5)
S += parametric_plot3d([t, y0, c*sqrt((y0/b)^2+(t/a)^2)], (t,-tx,tx), color="seagreen", thickness=1.5)
S += parametric_plot3d([t, y0, -c*sqrt((y0/b)^2+(t/a)^2)], (t,-tx,tx), color="seagreen", thickness=1.5)
xmax, ymax, zmax = a*vmax, b*vmax, c*vmax
S += arrow3d((0,0,0), (xmax+0.5,0,0), color="black")
S += text3d("x", (xmax+0.8,0,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,ymax+0.5,0), color="black")
S += text3d("y", (0,ymax+0.8,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,0,zmax+0.5), color="black")
S += text3d("z", (0,0,zmax+0.8), color="black", fontsize=26, frame = False)
S
```

### Figure C.2.7

Un paraboloïde elliptique.


```sage
var('r,theta,t')
a,b = 1.2, 1.8
rmax = 1.5
# paramétrage en coordonnées polaires (r,theta) : x=a*r*cos(theta), y=b*r*sin(theta),
# d'où z=(x/a)^2+(y/b)^2 = r^2, indépendant de theta
def macouleur(r,theta):
    return float(0.35 + 0.3*(r/rmax)^2)
S = parametric_plot3d([a*r*cos(theta), b*r*sin(theta), r^2], (r,0,rmax), (theta,0,2*pi), color=(macouleur, colormaps.YlOrBr), mesh=False, opacity=1, frame=False)
xmax, ymax, zmax = a*rmax, b*rmax, rmax^2
# trace horizontale (z = cste), à mi-hauteur : une ellipse
r0 = rmax/1.3
S += parametric_plot3d([a*r0*cos(theta), b*r0*sin(theta), r0^2], (theta,0,2*pi), color="firebrick", thickness=1.5)
# sections verticales dans les plans y=0 et x=0 (à travers le sommet) : des paraboles
S += parametric_plot3d([t, 0, (t/a)^2], (t,-xmax,xmax), color="seagreen", thickness=1.5)
S += parametric_plot3d([0, t, (t/b)^2], (t,-ymax,ymax), color="steelblue", thickness=1.5)
# sections décalées à x=x0 et y=y0 strictement positifs : x0, y0 sont les
# coordonnées d'un même point p0 = (x0,y0,r0^2) de la surface, à la hauteur r0
# de l'ellipse ci-dessus (theta0 arbitraire) — les trois traces se croisent donc
# toutes en p0. ty, tx sont calculés pour que ces courbes restent sur la
# surface tronquée (r au plus rmax), sans la dépasser
theta0 = pi/4
x0 = a*r0*cos(theta0)
y0 = b*r0*sin(theta0)
ty = b*sqrt(rmax^2-(x0/a)^2)
tx = a*sqrt(rmax^2-(y0/b)^2)
S += parametric_plot3d([x0, t, (x0/a)^2+(t/b)^2], (t,-ty,ty), color="steelblue", thickness=1.5)
S += parametric_plot3d([t, y0, (t/a)^2+(y0/b)^2], (t,-tx,tx), color="seagreen", thickness=1.5)
S += arrow3d((0,0,0), (xmax+0.5,0,0), color="black")
S += text3d("x", (xmax+0.8,0,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,ymax+0.5,0), color="black")
S += text3d("y", (0,ymax+0.8,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,0,zmax+0.5), color="black")
S += text3d("z", (0,0,zmax+0.8), color="black", fontsize=26, frame = False)
S
```

### Figure C.2.8

Un paraboloïde hyperbolique.


```sage
var('r,theta,t')
a,b = 1.2, 1.8
rmax = 1.5
# paramétrage en coordonnées polaires (r,theta) : x=a*r*cos(theta), y=b*r*sin(theta),
# d'où z=(x/a)^2-(y/b)^2 = r^2*cos(2*theta) (identité cos^2-sin^2=cos(2*theta))
def macouleur(r,theta):
    z = r^2*cos(2*theta)
    return float(0.5 + 0.25*(z/rmax^2))
S = parametric_plot3d([a*r*cos(theta), b*r*sin(theta), r^2*cos(2*theta)], (r,0,rmax), (theta,0,2*pi), color=(macouleur, colormaps.YlOrBr), mesh=False, opacity=1, frame=False)
xmax, ymax, zmax = a*rmax, b*rmax, rmax^2
# trace z=0 (cos(2*theta)=0 en theta=+-pi/4) : dégénère en deux droites x/a = +- y/b,
# qui se croisent au col
S += parametric_plot3d([a*r*cos(pi/4), b*r*sin(pi/4), 0], (r,-rmax,rmax), color="firebrick", thickness=1.5)
S += parametric_plot3d([a*r*cos(-pi/4), b*r*sin(-pi/4), 0], (r,-rmax,rmax), color="firebrick", thickness=1.5)
# sections verticales dans les plans y=0 et x=0 (à travers le col) : des paraboles
# de concavités opposées
S += parametric_plot3d([t, 0, (t/a)^2], (t,-xmax,xmax), color="seagreen", thickness=1.5)
S += parametric_plot3d([0, t, -(t/b)^2], (t,-ymax,ymax), color="steelblue", thickness=1.5)
# sections décalées à x=x0 et y=y0 strictement positifs ; ty, tx sont calculés
# pour que ces courbes restent sur la surface tronquée (r au plus rmax), sans la dépasser
x0, y0 = xmax/2, ymax/2
ty = b*sqrt(rmax^2-(x0/a)^2)
tx = a*sqrt(rmax^2-(y0/b)^2)
S += parametric_plot3d([x0, t, (x0/a)^2-(t/b)^2], (t,-ty,ty), color="steelblue", thickness=1.5)
S += parametric_plot3d([t, y0, (t/a)^2-(y0/b)^2], (t,-tx,tx), color="seagreen", thickness=1.5)
S += arrow3d((0,0,0), (xmax+0.5,0,0), color="black")
S += text3d("x", (xmax+0.8,0,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,ymax+0.5,0), color="black")
S += text3d("y", (0,ymax+0.8,0), color="black", fontsize=26)
S += arrow3d((0,0,0), (0,0,zmax+0.5), color="black")
S += text3d("z", (0,0,zmax+0.8), color="black", fontsize=26, frame = False)
S
```
