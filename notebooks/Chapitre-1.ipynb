# Chapitre 1 — code Sage des figures 3D

Code SageMath d'origine des figures 3D du chapitre 1, remplacées dans le manuel par des figures Asymptote.
Chaque cellule est autonome : on peut la copier dans [SageMathCell](https://sagecell.sagemath.org) pour obtenir une figure qu'on peut faire pivoter.

Les numéros de figures sont ceux de la version compilée au moment de l'extraction (octobre 2026).


## Section 1.5 — Courbes paramétriques et leurs tangentes


### Figure 1.5.9

L'intersection de l'ellipsoïde $x^2+\frac{y^2}{2}+\frac{z^2}{3}=1$ et du paraboloïde $z=x^2+2y^2$.


```sage
var('x,y,z')

f(x,y,z) = x^2+y^2/2+z^2/3
ellipsoide = implicit_plot3d(f==1,(x,-2,2),(y,-2,2),(z,0,3),texture ='orange',frame = False)

g(x,y) = x^2+2*y^2
paraboloide = implicit_plot3d(z==g,(x,-2,2),(y,-2,2),(z,0,2.5), opacity = 0.65, texture ='lightblue', frame = False)

h = solve([f==1 , z==g], x,y)
ep = 0.00000001
intersection = parametric_plot3d((x.subs(h[0][0]), y.subs(h[0][1]), z  ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep), thickness = 3,color='red')
intersection += parametric_plot3d( ( x.subs(h[1][0]), y.subs(h[1][1]), z ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep),thickness = 3,color = 'red')
intersection += parametric_plot3d( ( x.subs(h[2][0]), y.subs(h[2][1]), z ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep) ,thickness = 3, color = 'red')
intersection += parametric_plot3d( ( x.subs(h[3][0]), y.subs(h[3][1]), z ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep) ,thickness = 3,color='red')

S = ellipsoide + paraboloide + intersection

S += arrow3d((0,0,0),(1.6,0,0), color = "black", width= 0.5)
S += text3d("x", (1.7,0,0), color = "black", fontsize = 22)
S += arrow3d((0,0,0),(0,1.6,0), color = "black",  width= 0.5)
S += text3d("y", (0,1.7,0), color = "black", fontsize = 22)
S += arrow3d((0,0,0),(0,0,3), color = "black", width = 0.5)
S += text3d("z", (0,0,3.1), color = "black", fontsize = 22)
S
```

### Figure 1.5.11

Courbe paramétrée par $\vec r(y) = e^y\,\vec\imath + y\,\vec\jmath + (e^y-e^{2y})\,\vec k$, intersection des surfaces $x^3=e^{3y}$ et $x^2-e^y+z=0$.


```sage
var('x,y,z')
f(x,y,z) = x^3-exp(3*y)
g(x,y,z) = x^2-exp(y)+z
h = solve ([f == 0,g == 0],x,z)
ellipsoide = implicit_plot3d(f,(x,-2,2),(y,-2,2),(z,-2,1),opacity = 0.65,texture ='orange',frame = False)
paraboloide = implicit_plot3d(g,(x,-2,2),(y,-2,2),(z,-2,1), opacity = 0.65, texture ='lightblue', frame = False)
S = parametric_plot3d((x.subs(h[2][0]),y,z.subs(h[2][1])), (y,-2,0.70), thickness = 3,color='red')
S += ellipsoide
S += paraboloide
S += arrow3d((0,0,0),(1.6,0,0), color = "black", width= 0.5)
S += text3d("x", (1.7,0,0), color = "black", fontsize = 22)
S += arrow3d((0,0,0),(0,1.6,0), color = "black",  width= 0.5)
S += text3d("y", (0,1.7,0), color = "black", fontsize = 22)
S += arrow3d((0,0,0),(0,0,1.2), color = "black", width = 0.5)
S += text3d("z", (0,0,1.3), color = "black", fontsize = 22)
S
```

### Figure 1.5.13

Quatre portions de la courbe d'intersection précédente, données par les différents signes de $x$ et de $y$.


```sage
var('x,y,z')
f(x,y,z) = x^2+y^2/2+z^2/3
g(x,y,z) = x^2+2*y^2 - z
h = solve ([f == 1,g == 0],x,y)
ep = 0.0000001
ellipsoide = implicit_plot3d(f==1,(x,-2,2),(y,-2,2),(z,0,2),texture ='orange',frame = False)
paraboloide = implicit_plot3d(g,(x,-2,2),(y,-2,2),(z,0,2), opacity = 0.65, texture ='lightblue', frame = False)
S = parametric_plot3d((x.subs(h[0][0]), y.subs(h[0][1]), z  ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep), thickness = 5,color='red')
S += parametric_plot3d( ( x.subs(h[1][0]), y.subs(h[1][1]), z ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep),thickness = 5,color = 'blue')
S += parametric_plot3d( ( x.subs(h[2][0]), y.subs(h[2][1]), z ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep) ,thickness = 5, color = 'green')
S += parametric_plot3d( ( x.subs(h[3][0]), y.subs(h[3][1]), z ), (z,-3/2 + sqrt(21)/2 +ep,-3/8 + sqrt(201)/8-ep) ,thickness = 5,color='black')
S += ellipsoide
S += paraboloide
S += arrow3d((0,0,0),(1.6,0,0), color = "black", width= 0.5)
S += text3d("x", (1.7,0,0), color = "black", fontsize = 22)
S += arrow3d((0,0,0),(0,1.6,0), color = "black",  width= 0.5)
S += text3d("y", (0,1.7,0), color = "black", fontsize = 22)
S += arrow3d((0,0,0),(0,0,3), color = "black", width = 0.5)
S += text3d("z", (0,0,3.1), color = "black", fontsize = 22)
S
```

### Figure 1.5.15

Le plan $z = 1+y$ et le cône $z^2 = x^2+y^2$, et leur intersection.


```sage
var('u,v,r,t')
Fig = parametric_plot3d([u,v,1+v],(u,-2,2),(v,-1,1.6), mesh = True,
  texture = "lightblue", frame = False)
Fig += parametric_plot3d([r*cos(t), r*sin(t),r], (r,0, 2),(t,0,2*pi), mesh = True, texture = "orange", opacity = 0.75)
Fig += parametric_plot3d([r,(r^2 - 1)/2,(r^2 + 1)/2], (r,-2,2),
  color = "red", thickness = 5)
Fig
```
