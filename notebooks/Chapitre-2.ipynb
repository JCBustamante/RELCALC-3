## Section 2.1 — Fonctions de plusieurs variables


### Figure 2.1.9

Une surface $z=f(x,y)$ ainsi que quelques courbes de niveau et quelques traces horizontales.


```sage
var('x y z')
g(x,y,z) = 2+2*(1-x)^2*exp(-(x^2) - (y+1)^2) +2*abs(x/5 - x^3 - y^5)*exp(-x^2-y^2)+ 1*exp(-(x+1)^2 - y^2)
f = z==g
S = plot3d(g(x,y), (x,-2,2),(y,-2,2), mesh=0, color = "orange", opacity = 0.5, frame = False, projection = "orthographic")
levels=srange(2,7,0.5)
Contours = Graphics()
for zlevel in levels:
        S+=implicit_plot3d(g(x,y,z)-zlevel, (x,-2,2),(y,-2,2),(z,zlevel - 0.02,zlevel + 0.02), color = "brown")
for zlevel in levels:
        Contours+=implicit_plot3d(f(x,y,zlevel), (x,-2,2),(y,-2,2),(z,-     0.03,+0.03),thickness=3, color = "blue")
S +=Contours
S += arrow3d((0,0,0),(2,0,0), color = "black")
S += text3d("x", (2.2,0,0), color = "black", fontsize = 24)
S += arrow3d((0,0,0),(0,3,0), color = "black")
S += text3d("y", (0,3.2,0), color = "black", fontsize = 24)
S += arrow3d((0,0,0),(0,0,7), color = "black")
S += text3d("z", (0,0,7.2), color = "black", fontsize = 24)
S += plot3d(0, (x,-2,2),(y,-2,2), color = "blue", opacity=0.25)
S
```

## Section 2.2 — Limites et continuité

Dans le code d'origine, l'option `région=` (avec un accent) était ignorée par Sage ; elle est corrigée ici en `region=`.


### Figure 2.2.4

Le graphe de $f(x,y)=\frac{x^2-y^2}{x^2+y^2}$.


```sage
var('x,y,z')
f(x,y) = (x^2-y^2)/(x^2 + y^2)
cm = colormaps.Spectral
zmin, zmax = -1, 1
def c(x,y,z):
    return float((f(x,y)-zmin)/(zmax-zmin))
fig = implicit_plot3d(z==f, (x,-2,2), (y,-2,2), (z,zmin,zmax), color = (c,cm),region=lambda x,y,z: x^2+y^2+z^2>0.00001, frame = False, plot_points = 70)
fig
```

## Section 2.4 — Différentiabilité et approximations linéaires

## Section 2.8 — Valeurs maximales et minimales


### Figure 2.8.17

Le graphe de $f(x,y)=x^2-y^2$, avec ses traces en $x=0$ (bleu) et $y=0$ (rouge).


```sage
var('x,y')
f(x,y) = x^2 - y^2
def macouleur(x,y):
    return float((f(x,y)+4)/8)
S = plot3d(f(x,y), (x,-2,2), (y,-2,2), color=(macouleur,colormaps.Spectral), mesh = True, opacity = 0.85, frame = False)
S += parametric_plot3d([x,0,x^2], (x,-2,2), color = "red", thickness = 4)
S += parametric_plot3d([0,y,-y^2], (y,-2,2), color = "blue", thickness = 4)
S += point3d((0,0,0), color = "black", size = 40)
S += arrow3d((0,0,0),(2.3,0,0), color = "black")
S += text3d("x", (2.5,0,0), color = "black", fontsize = 20)
S += arrow3d((0,0,0),(0,2.3,0), color = "black")
S += text3d("y", (0,2.5,0), color = "black", fontsize = 20)
S += arrow3d((0,0,0),(0,0,4.5), color = "black")
S += text3d("z", (0,0,4.7), color = "black", fontsize = 20)
S
```
