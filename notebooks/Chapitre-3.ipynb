# Chapitre 3 — code Sage des figures 3D

Code SageMath d'origine des figures 3D du chapitre 3, remplacées dans le manuel par des figures Asymptote.
Chaque cellule est autonome : on peut la copier dans [SageMathCell](https://sagecell.sagemath.org) pour obtenir une figure qu'on peut faire pivoter.

Les numéros de figures sont ceux de la version compilée au moment de l'extraction (octobre 2026).


## Section 3.4 — Intégrales triples


### Figure 3.4.8

La région $\mathcal{E}$ de l'exemple : sous le plan $y+z=1$, au-dessus du plan $Oxy$, à l'intérieur du cylindre parabolique $y=x^2$.


```sage
var('x,y,z')
Base = parametric_plot3d([(1-y)*x+y,(1-y)*x^2 + y,0],(x,-1,1), (y,0,1), color = "orange", opacity = 0.35,frame = False, projection = 'orthographic')
Cyl = parametric_plot3d([x,x^2, (1-y)*(1-x^2)],(x,-1,1),(y,0,1), color = "green", opacity = 0.35,frame = False)
Plan = parametric_plot3d([(1-y)*x+y,(1-y)*x^2 + y,1-(1-y)*x^2 - y ],(x,-1,1), (y,0,1), color = "blue", opacity = 0.35,frame = False)
Projxy = parametric_plot3d([x,0, (1-y)*(1-x^2)],(x,-1,1),(y,0,1), color = "green", opacity = 0.25,frame = False)
Fig = Base + Cyl + Plan+ Projxy

Fig += parametric_plot3d([x,x^2,0], (x,-1,1), color = "darkblue", thickness = 2)
Fig += parametric_plot3d([x,1,0], (x,-1,1), color = "darkblue", thickness = 2)
Fig += parametric_plot3d([x,x^2,1-x^2], (x,-1,1), color = "darkblue", thickness = 2)

Fig += arrow3d((0,0,0),(1.15,0,0), width=0.5, color = "black",frame = False)
Fig += text3d("x", (1.25,0,0), color = "black",frame = False)
Fig += arrow3d((0,0,0),(0,1.15,0), width=0.5, color = "black",frame = False)
Fig += text3d("y", (0,1.25,0), color = "black",frame = False)
Fig += arrow3d((0,0,0),(0,0,1.15), width=0.5, color = "black",frame = False)
Fig += text3d("z", (0,0,1.25), color = "black",frame = False)
Fig
```

### Exercice 5 de la section 3.4

La région du premier octant limitée par les plans $x=0$, $y=0$, $z=0$, $y+z=2$ et le cylindre parabolique $x=4-y^2$.


```sage
var('s,t')
Fig =  parametric_plot3d((1-s)*vector([0,t,0]) + s*vector([4-t^2,t,0]), (s,0,1), (t,0,2), color = "orange", opacity = 0.55)
Fig +=  parametric_plot3d((1-s)*vector([0,t,0]) + s*vector([4-t^2,t,0]) + vector([0,0,2-t]), (s,0,1), (t,0,2), color = "crimson", opacity = 0.25)
Fig += parametric_plot3d([4-t^2, t, s*(2-t)], (t,0,2), (s,0,1), color = "steelblue", opacity = 0.65)
Fig += parametric_plot3d([0,t,s*(2-t)], (t,0,2), (s,0,1), color = "grey", opacity = 0.65)
Fig += parametric_plot3d([s,0,t], (s,0,4), (t,0,2), color = "green", opacity = 0.25)

Fig += parametric_plot3d([4-t^2,t,0], (t,0,2), color = "darkblue", thickness= 2)
Fig += parametric_plot3d([4-t^2,t,2-t], (t,0,2), color = "darkblue", thickness= 2)
Fig += parametric_plot3d([0,t,2-t], (t,0,2), color = "darkblue", thickness= 2)
Fig += parametric_plot3d([t,0,2], (t,0,4), color = "darkblue", thickness= 2)
Fig += parametric_plot3d([4,0,t], (t,0,2), color = "darkblue", thickness= 2)


Fig += arrow3d((0,0,0),(4.2,0,0), width=0.75, color = "black",frame = False)
Fig += text3d("x", (4.4,0,0), color = "black",frame = False, fontsize = 26)
Fig += arrow3d((0,0,0),(0,2.2,0), width=0.75, color = "black",frame = False)
Fig += text3d("y", (0,2.4,0), color = "black",frame = False, fontsize = 26)
Fig += arrow3d((0,0,0),(0,0,2.2), width=0.75, color = "black",frame = False)
Fig += text3d("z", (0,0,2.4), color = "black", projection ="orthographic", fontsize = 26)
Fig
```
