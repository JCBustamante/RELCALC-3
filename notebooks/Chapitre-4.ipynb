## Section 4.7 — Intégrales de surface


### Figure 4.7.16

Le ruban de Möbius et son vecteur normal.


```sage
var('s,t')
R = 2
def x(s,t):
    return (R+s*cos(t/2))*cos(t)
def y(s,t):
    return (R+s*cos(t/2))*sin(t)
def z(s,t):
    return s*sin(t/2)
def macouleur(s,t):
    return (1+sin(t))/2
S = parametric_plot3d((x(s,t), y(s,t), z(s,t)), (s,-1,1), (t,0,2*pi), color=(macouleur, colormaps.coolwarm), frame = False)
Bord = parametric_plot3d((x(-1,t), y(-1,t), z(-1,t)), (t,0,2*pi), color="black", thickness=3)
L = 0.6
Normales = sum(arrow((x(0,t), y(0,t), z(0,t)),
        (x(0,t) - L*cos(t)*sin(t/2), y(0,t) - L*sin(t)*sin(t/2), z(0,t) + L*cos(t/2)),
        color="black", width=0.5) for t in sxrange(0,2*pi,0.3))
S + Bord + Normales
```

### Figure 4.9.16

Les surfaces $\mathcal{S}$ et $\mathcal{S}'$, avec leurs vecteurs normaux.


```sage
var('r,t')
X(r,t) = r*cos(t)
Y(r,t) = r*sin(t)
f(x,y) = (1-x^2 -y^2)*(1-y^3)*cos(x)*exp(y)
fx = diff(f,x)
fy = diff(f,y)
x0, y0 = 0.5,0.3
h=2*sqrt(fx(x0,y0)^2 + fy(x0,y0)^2 +1)
Fig = parametric_plot3d([X(r,t), Y(r,t), f(X(r,t), Y(r,t)) ],
                        (r,0,1), (t,0,2*pi), color = "steelblue", mesh = True,opacity = 0.65)
Fig += parametric_plot3d([X(r,t), Y(r,t), 0 ],
                        (r,0,1), (t,0,2*pi), color = "firebrick", mesh = False)
Fig+= arrow3d((0.5,0.3,0),(0.5,0.3,-0.5), width=1, color = "firebrick",frame = False)
Fig+= arrow3d((x0,y0,f(x0,y0)),(x0-fx(x0,y0)/h,y0 - fy(x0,y0)/h,
  f(x0,y0) + 1/h), width=1, color = "steelblue",frame = False)
Fig += arrow3d((0,0,0),(1.5,0,0), width=0.7, color = "black",
  frame = False)
Fig += text3d("x", (1.7,0,0), color = "black",frame = False, fontsize = 20)
Fig += arrow3d((0,0,0),(0,1.5,0), width=0.7, color = "black",
  frame = False)
Fig += text3d("y", (0,1.7,0), color = "black",frame = False, fontsize = 20)
Fig += arrow3d((0,0,0),(0,0,1.5), width=0.7, color = "black",
  frame = False)
Fig += text3d("z", (0,0,1.7), color = "black", fontsize = 20)
Fig
```
